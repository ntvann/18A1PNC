```python
import requests

def doc_bai_post():
    url = "https://jsonplaceholder.typicode.com/posts"
    
    try:
        response = requests.get(url)
        response.raise_for_status()
        
        posts = response.json()
        
        print(f"Tổng số bài post: {len(posts)}")
        print("\nDanh sách các bài post:")
        print("=" * 80)
        
        for i, post in enumerate(posts[:10], 1):  # Hiển thị 10 bài đầu
            print(f"Bài {i}:")
            print(f"  UserID: {post['userId']}")
            print(f"  ID: {post['id']}")
            print(f"  Tiêu đề: {post['title']}")
            print(f"  Nội dung: {post['body'][:100]}...")  # Giới hạn nội dung
            print("-" * 80)
            
    except Exception as e:
        print(f"Lỗi khi đọc API: {e}")

doc_bai_post()
```
