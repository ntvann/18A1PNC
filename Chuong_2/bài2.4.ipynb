```python
from xml.dom import minidom
import os

def tai_va_phan_tich_xml():
    try:
        print("BÀI 2.4: TẢI VÀ PHÂN TÍCH FILE sample.xml BẰNG MINIDOM")
        print("=" * 60)

        xml_file = "sample.xml"

        # Kiểm tra sự tồn tại của file
        if not os.path.exists(xml_file):
            print("Lỗi: Không tìm thấy file sample.xml trong thư mục hiện tại.")
            return None
        
        # Đọc và phân tích file XML
        print("Đang tải file sample.xml ...")
        doc = minidom.parse(xml_file)
        print("Đã tải file XML thành công!\n")

        # Lấy phần tử gốc
        root = doc.documentElement
        print(f"Phần tử gốc: {root.tagName}")
        print(f"Tổng số phần tử con trực tiếp: {len(root.childNodes)}\n")

        # Hàm đệ quy in cấu trúc XML dạng cây
        def phan_tich_node(node, level=0):
            indent = "  " * level
            if node.nodeType == node.ELEMENT_NODE:
                print(f"{indent}<{node.tagName}>")
                for child in node.childNodes:
                    phan_tich_node(child, level + 1)
                print(f"{indent}</{node.tagName}>")
            elif node.nodeType == node.TEXT_NODE:
                text = node.data.strip()
                if text:
                    print(f"{indent}{text}")

        print("CẤU TRÚC CÂY XML:")
        print("-" * 30)
        phan_tich_node(root)
        print("-" * 30)

        # Thống kê các phần tử trong file
        all_elements = doc.getElementsByTagName('*')
        element_count = {}
        for el in all_elements:
            tag = el.tagName
            element_count[tag] = element_count.get(tag, 0) + 1

        print("\nTHỐNG KÊ PHẦN TỬ TRONG FILE sample.xml:")
        for tag, count in element_count.items():
            print(f"  <{tag}> : {count} phần tử")

        print(f"Tổng số phần tử: {len(all_elements)}\n")

        # Lấy thông tin tên công ty
        name_nodes = doc.getElementsByTagName('name')
        if name_nodes:
            print(f"Tên công ty: {name_nodes[0].firstChild.data}")

        return doc

    except Exception as e:
        print(f"Lỗi khi phân tích XML: {e}")
        return None


# ----------------------------------------
# GỌI HÀM THỰC HIỆN PHÂN TÍCH FILE sample.xml
# ----------------------------------------
doc = tai_va_phan_tich_xml()

# ----------------------------------------
# KIỂM TRA FILE ĐÃ TẠO VÀ HIỂN THỊ NỘI DUNG
# ----------------------------------------
if os.path.exists("sample.xml"):
    print("\nNỘI DUNG FILE sample.xml:")
    print("=" * 30)
    with open("sample.xml", "r", encoding="utf-8") as f:
        print(f.read())
else:
    print("File sample.xml không tồn tại.")
```
