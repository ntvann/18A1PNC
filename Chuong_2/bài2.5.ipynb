```python
from xml.dom import minidom
import os

def lay_danh_sach_phan_tu():
    try:
        print("BÀI 2.5: SỬ DỤNG getElementsByTagName() ĐỂ LẤY DANH SÁCH PHẦN TỬ")
        print("=" * 70)

        # Tải file XML
        xml_file = "sample.xml"
        if not os.path.exists(xml_file):
            print("Lỗi: Không tìm thấy file sample.xml trong thư mục hiện tại.")
            return None

        doc = minidom.parse(xml_file)
        print("Đã tải file sample.xml thành công!\n")

        # Lấy tất cả các phần tử trong tài liệu
        all_elements = doc.getElementsByTagName('*')
        print("DANH SÁCH TẤT CẢ CÁC PHẦN TỬ:")
        print("-" * 50)
        for i, element in enumerate(all_elements, 1):
            print(f"{i}. <{element.tagName}>")

            # Hiển thị thuộc tính (nếu có)
            if element.attributes:
                for attr_name, attr_value in element.attributes.items():
                    print(f"   Thuộc tính: {attr_name} = {attr_value}")

            # Hiển thị nội dung (nếu có)
            if element.firstChild and element.firstChild.nodeType == element.TEXT_NODE:
                content = element.firstChild.data.strip()
                if content:
                    print(f"   Nội dung: {content}")

            # Liệt kê các phần tử con (nếu có)
            child_elements = [
                child for child in element.childNodes if child.nodeType == child.ELEMENT_NODE
            ]
            if child_elements:
                child_names = [child.tagName for child in child_elements]
                print(f"   Phần tử con: {', '.join(child_names)}")

            print()

        # --------------------------------------------------------
        # LẤY DANH SÁCH THEO TÊN THẺ
        # --------------------------------------------------------
        print("\nLẤY DANH SÁCH CỤ THỂ THEO TÊN THẺ")
        print("-" * 50)

        # Các phần tử <name>
        name_elements = doc.getElementsByTagName('name')
        print(f"Tìm thấy {len(name_elements)} phần tử <name>:")
        for i, name_element in enumerate(name_elements, 1):
            content = name_element.firstChild.data.strip() if name_element.firstChild else "Không có nội dung"
            print(f"  {i}. {content}")

        # Các phần tử <staff>
        staff_elements = doc.getElementsByTagName('staff')
        print(f"\nTìm thấy {len(staff_elements)} phần tử <staff>:")
        for i, staff_element in enumerate(staff_elements, 1):
            staff_id = staff_element.getAttribute('id')
            print(f"  {i}. Mã nhân viên: {staff_id}")

            name = staff_element.getElementsByTagName('name')[0].firstChild.data
            salary = staff_element.getElementsByTagName('salary')[0].firstChild.data
            print(f"     Họ tên: {name}")
            print(f"     Lương: {salary}")

        # Các phần tử <salary>
        salary_elements = doc.getElementsByTagName('salary')
        print(f"\nTìm thấy {len(salary_elements)} phần tử <salary>:")
        for i, salary_element in enumerate(salary_elements, 1):
            content = salary_element.firstChild.data.strip() if salary_element.firstChild else "Không có nội dung"
            print(f"  {i}. {content}")

        # --------------------------------------------------------
        # THỐNG KÊ TỔNG QUAN
        # --------------------------------------------------------
        print("\nTHỐNG KÊ TỔNG QUAN:")
        print("-" * 25)
        element_stats = {}
        for element in all_elements:
            tag = element.tagName
            element_stats[tag] = element_stats.get(tag, 0) + 1

        for tag_name, count in element_stats.items():
            print(f"  <{tag_name}> : {count} phần tử")

        print(f"\nTổng số phần tử trong file: {len(all_elements)}")

        

        # Phần tử <company>
        company_elements = doc.getElementsByTagName('company')
        if company_elements:
            company = company_elements[0]
            print(f"Phần tử gốc: <{company.tagName}>")

        # Thử tìm phần tử không tồn tại
        nonexistent = doc.getElementsByTagName('department')
        print(f"Số phần tử <department> (không tồn tại): {len(nonexistent)}")

        return doc

    except Exception as e:
        print(f"Lỗi khi phân tích XML: {e}")
        return None


# ----------------------------------------
# GỌI HÀM THỰC HIỆN
# ----------------------------------------
doc = lay_danh_sach_phan_tu()

# ----------------------------------------
# KIỂM TRA FILE ĐÃ TỒN TẠI
# ----------------------------------------
print("\nKIỂM TRA FILE:")
print("=" * 30)
if os.path.exists('sample.xml'):
    file_size = os.path.getsize('sample.xml')
    print(f"sample.xml tồn tại ({file_size} bytes)")
else:
    print("Không tìm thấy file sample.xml.")
```
