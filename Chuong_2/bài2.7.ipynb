```python
import json
def json_to_python():
    json_data = '{"name": "Nguyen Van A", "age": 21, "city": "Ha Noi", "active": true}'
    
    python_obj = json.loads(json_data)
    
    print("Python object:")
    print(python_obj)
    print(f"Kieu du lieu: {type(python_obj)}")
    
    return python_obj

# Chạy chương trình
result = json_to_python()
```
