```python
import json

def python_to_json():
    python_obj = {
        "name": "Tran Thi B",
        "age": 22,
        "city": "TP.HCM",
        "courses": ["Python", "Java"]
    }
    
    json_string = json.dumps(python_obj, ensure_ascii=False)
    print("Chuoi JSON:")
    print(json_string)
    
    return json_string

result = python_to_json()
```
