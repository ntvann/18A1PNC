```python
import requests
import xml.etree.ElementTree as ET
import csv
import os

def rss_to_csv():
    """
    Bài 2.6: Tải RSS feed, phân tích XML và lưu vào file CSV
    """
    url = "https://www.hindustantimes.com/rss/topnews/rssfeed.xml"
    
    try:
        print("BAI 2.6: TAI RSS FEED VA LUU VAO CSV")
        print("=" * 50)
        
        print("Dang tai RSS feed...")
        
        # Tải RSS feed
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
        }
        response = requests.get(url, headers=headers, timeout=10)
        response.raise_for_status()
        
        # Lưu file XML
        with open('rss_feed.xml', 'w', encoding='utf-8') as f:
            f.write(response.text)
        print("Da luu RSS feed thanh file rss_feed.xml")
        
        # Phân tích XML
        root = ET.fromstring(response.content)
        
        # Tạo danh sách tin tức
        news_list = []
        
        # Tìm tất cả các item (tin tức)
        print("Dang phan tich tin tuc...")
        for item in root.findall('.//item'):
            news_item = {}
            
            # Lấy các thông tin cơ bản
            for child in item:
                tag = child.tag.split('}')[-1] if '}' in child.tag else child.tag
                text = child.text.strip() if child.text else ''
                news_item[tag] = text
            
            news_list.append(news_item)
        
        # Lưu vào CSV
        if news_list:
            csv_filename = 'news.csv'
            with open(csv_filename, 'w', newline='', encoding='utf-8') as csvfile:
                fieldnames = news_list[0].keys()
                writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
                
                writer.writeheader()
                for news in news_list:
                    writer.writerow(news)
            
            print(f"Da luu {len(news_list)} tin tuc vao file {csv_filename}")
            
            # Hiển thị thông tin về file CSV đã lưu
            print("\nTHONG TIN FILE CSV DA LUU:")
            print("-" * 30)
            if os.path.exists(csv_filename):
                file_size = os.path.getsize(csv_filename)
                print(f"Ten file: {csv_filename}")
                print(f"Kich thuoc: {file_size} bytes")
                print(f"So dong tin tuc: {len(news_list)}")
                
                # Hiển thị 5 dòng đầu tiên từ file CSV
                print(f"\n5 DONG DAU TIEN TRONG FILE {csv_filename}:")
                print("-" * 40)
                with open(csv_filename, 'r', encoding='utf-8') as f:
                    lines = f.readlines()[:6]  # 5 dòng dữ liệu + header
                    for i, line in enumerate(lines):
                        if i == 0:
                            print(f"HEADER: {line.strip()}")
                        else:
                            print(f"Dong {i}: {line.strip()}")
            
            # Hiển thị 3 tin đầu tiên chi tiết
            print("\n3 TIN TUC DAU TIEN:")
            print("-" * 25)
            for i, news in enumerate(news_list[:3], 1):
                print(f"{i}. Tieu de: {news.get('title', 'Khong co tieu de')}")
                print(f"   Ngay dang: {news.get('pubDate', 'Khong co ngay')}")
                print(f"   Link: {news.get('link', 'Khong co link')}")
                print(f"   Mo ta: {news.get('description', 'Khong co mo ta')[:100]}...")
                print()
                
        else:
            print("Khong tim thay tin tuc nao")
            
        # Hiển thị thống kê
        print("\nTHONG KE:")
        print("-" * 15)
        print(f"Tong so tin tuc: {len(news_list)}")
        
        if news_list:
            print(f"So cot trong CSV: {len(news_list[0].keys())}")
            print(f"Cac cot: {', '.join(news_list[0].keys())}")
            
    except Exception as e:
        print(f"Loi: {e}")

# Gọi hàm thực hiện
rss_to_csv()

# Kiểm tra các file đã tạo
print("\nKIEM TRA CAC FILE DA TAO:")
print("=" * 25)

files_to_check = ['rss_feed.xml', 'news.csv']

for file in files_to_check:
    if os.path.exists(file):
        file_size = os.path.getsize(file)
        print(f"{file}: Ton tai ({file_size} bytes)")
        
        # Hiển thị thông tin thêm về file CSV
        if file == 'news.csv':
            with open(file, 'r', encoding='utf-8') as f:
                lines = f.readlines()
                print(f"  So dong: {len(lines)} (1 header + {len(lines)-1} du lieu)")
    else:
        print(f"{file}: Khong ton tai")

# Hiển thị đường dẫn đầy đủ đến file CSV
print(f"\nDuong dan day du den file CSV:")
print(f"{os.path.abspath('news.csv')}")
```
