```python
import json
import datetime
import os

class QuanLyGiaoDich:
    def __init__(self):
        self.danh_sach_giao_dich = []
    
    def them_giao_dich_tien(self, ma, ngay, don_gia, so_luong, loai):
        giao_dich = {
            "ma": ma,
            "ngay": ngay,
            "don_gia": don_gia,
            "so_luong": so_luong,
            "loai": loai,
            "loai_giao_dich": "tien"
        }
        self.danh_sach_giao_dich.append(giao_dich)
        print(f"Đã thêm giao dịch tiền: {ma}")
    
    def them_giao_dich_vang(self, ma, ngay, don_gia, so_luong, loai_vang):
        giao_dich = {
            "ma": ma,
            "ngay": ngay,
            "don_gia": don_gia,
            "so_luong": so_luong,
            "loai_vang": loai_vang,
            "loai_giao_dich": "vang"
        }
        self.danh_sach_giao_dich.append(giao_dich)
        print(f"Đã thêm giao dịch vàng: {ma}")
    
    def luu_file_json(self):
        if not self.danh_sach_giao_dich:
            print("Không có giao dịch nào để lưu!")
            return None
            
        # Tạo tên file theo định dạng: ngay-thang-nam-gio-phut-giay.json
        thoi_gian = datetime.datetime.now().strftime("%d-%m-%Y-%H-%M-%S")
        ten_file = f"{thoi_gian}.json"
        
        try:
            with open(ten_file, 'w', encoding='utf-8') as f:
                json.dump(self.danh_sach_giao_dich, f, ensure_ascii=False, indent=2)
            
            print(f"Đã lưu {len(self.danh_sach_giao_dich)} giao dịch vào file: {ten_file}")
            print(f"Đường dẫn: {os.path.abspath(ten_file)}")
            return ten_file
        except Exception as e:
            print(f"Lỗi khi lưu file: {e}")
            return None


def main():
    ql = QuanLyGiaoDich()
    print("=== CHƯƠNG TRÌNH QUẢN LÝ GIAO DỊCH ===")
    
    while True:
        print("\n1. Thêm giao dịch tiền")
        print("2. Thêm giao dịch vàng")
        print("3. Xem danh sách giao dịch")
        print("4. Kết thúc và lưu file JSON")
        
        choice = input("Chọn chức năng (1-4): ")
        
        if choice == '1':
            print("\n--- THÊM GIAO DỊCH TIỀN ---")
            ma = input("Nhập mã giao dịch: ")
            ngay = input("Nhập ngày giao dịch (dd/mm/yyyy): ")
            don_gia = float(input("Nhập đơn giá: "))
            so_luong = int(input("Nhập số lượng: "))
            loai = input("Nhập loại tiền (USD/EUR/VND): ")
            ql.them_giao_dich_tien(ma, ngay, don_gia, so_luong, loai)
            
        elif choice == '2':
            print("\n--- THÊM GIAO DỊCH VÀNG ---")
            ma = input("Nhập mã giao dịch: ")
            ngay = input("Nhập ngày giao dịch (dd/mm/yyyy): ")
            don_gia = float(input("Nhập đơn giá: "))
            so_luong = int(input("Nhập số lượng: "))
            loai_vang = input("Nhập loại vàng (SJC/24K/18K): ")
            ql.them_giao_dich_vang(ma, ngay, don_gia, so_luong, loai_vang)
            
        elif choice == '3':
            print(f"\n--- DANH SÁCH GIAO DỊCH ({len(ql.danh_sach_giao_dich)}) ---")
            if not ql.danh_sach_giao_dich:
                print("Chưa có giao dịch nào.")
            else:
                for gd in ql.danh_sach_giao_dich:
                    if gd["loai_giao_dich"] == "tien":
                        print(f"Mã: {gd['ma']} | Ngày: {gd['ngay']} | Tiền: {gd['loai']} | Đơn giá: {gd['don_gia']} | SL: {gd['so_luong']}")
                    else:
                        print(f"Mã: {gd['ma']} | Ngày: {gd['ngay']} | Vàng: {gd['loai_vang']} | Đơn giá: {gd['don_gia']} | SL: {gd['so_luong']}")
            
        elif choice == '4':
            if ql.danh_sach_giao_dich:
                luu = input("Bạn có muốn ghi vào tập tin không? (1: Có, 0: Không): ")
                if luu == '1':
                    ten_file = ql.luu_file_json()
                    if ten_file:
                        print("Lưu file thành công.")
                else:
                    print("Không lưu file.")
            else:
                print("Không có giao dịch nào để lưu.")
            break
            
        else:
            print("Lựa chọn không hợp lệ. Vui lòng chọn 1–4.")


if __name__ == "__main__":
    main()

# Kiểm tra file JSON đã tạo
print("\n" + "="*50)
print("KIỂM TRA FILE JSON TRONG THƯ MỤC:")
print("="*50)

json_files = [f for f in os.listdir('.') if f.endswith('.json')]
if json_files:
    print("Các file JSON đã tạo:")
    for file in json_files:
        file_size = os.path.getsize(file)
        print(f"   {file} ({file_size} bytes)")
else:
    print("Chưa có file JSON nào được tạo.")
```
