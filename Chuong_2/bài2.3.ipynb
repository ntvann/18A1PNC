```python
# Nội dung XML dựa trên yêu cầu của Bài 2.3 (thông tin nhân viên)
xml_content = """<?xml version="1.0"?>
<company>
    <name>GeeksForGeeks Company</name>
    <staff id="1">
        <name>Amar Pandey</name>
        <salary>8.5 LPA</salary>
    </staff>
    <staff id="2">
        <name>Akbhar Khan</name>
        <salary>6.5 LPA</salary>
    </staff>
    <staff id="3">
        <name>Anthony Walter</name>
        <salary>3.2 LPA</salary>
    </staff>
</company>
"""
file_name = "sample.xml"

# Ghi nội dung vào file
with open(file_name, "w", encoding="utf-8") as f:
    f.write(xml_content)

print(f"Đã tạo file {file_name} với nội dung XML (Bài 2.3) trong thư mục hiện tại.")
```
