```python
import requests

def doc_bai_post_noi_bat():
    url = "https://jsonplaceholder.typicode.com/comments?postId=1"
    
    try:
        response = requests.get(url)
        response.raise_for_status()
        
        comments = response.json()
        
        print("Danh sách các bài post nổi bật (postId = 1)")
        print("=" * 60)
        print(f"Tổng số comment: {len(comments)}")
        print("\nDanh sách 3 comment đầu tiên:")
        print("=" * 60)
        
        for i, comment in enumerate(comments[:3], 1):
            print(f"Comment {i}:")
            print(f"  ID: {comment['id']}")
            print(f"  Tên: {comment['name']}")
            print(f"  Email: {comment['email']}")
            print(f"  Nội dung: {comment['body'][:100]}...")
            print("-" * 60)
            
    except Exception as e:
        print(f"Lỗi khi đọc API: {e}")

doc_bai_post_noi_bat()
```
