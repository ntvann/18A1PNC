```python
import json

def thong_ke_nhan_vien():
    data = {
        "ten_cong_ty": "Công ty TNHH Đất Việt",
        "dia_chi": "abc Giải Phóng – Hà Nội",
        "nhan_vien": [
            {"ten": "Nguyễn Văn A", "don_vi": "Đơn vị A1"},
            {"ten": "Trần Thị B", "don_vi": "Đơn vị A2"},
            {"ten": "Lê Văn C", "don_vi": "Đơn vị A1"},
            {"ten": "Phạm Thị D", "don_vi": "Đơn vị A3"},
            {"ten": "Hoàng Văn E", "don_vi": "Đơn vị A2"},
            {"ten": "Vũ Thị F", "don_vi": "Đơn vị A4"},
            {"ten": "Đặng Văn G", "don_vi": "Đơn vị A3"},
            {"ten": "Bùi Thị H", "don_vi": "Đơn vị A4"},
            {"ten": "Mai Văn I", "don_vi": "Đơn vị A1"},
            {"ten": "Ngô Thị K", "don_vi": "Đơn vị A4"}
        ]
    }
    
    # Thống kê
    ten_cong_ty = data["ten_cong_ty"]
    dia_chi = data["dia_chi"]
    tong_nhan_vien = len(data["nhan_vien"])
    
    print(f"Tên công ty: {ten_cong_ty}")
    print(f"Địa chỉ: {dia_chi}.")
    print("---Thống kê nhân viên theo đơn vị---")
    
    # Thống kê theo đơn vị
    thong_ke = {}
    for nv in data["nhan_vien"]:
        don_vi = nv["don_vi"]
        thong_ke[don_vi] = thong_ke.get(don_vi, 0) + 1
    
    # Hiển thị kết quả
    for i, (don_vi, so_luong) in enumerate(thong_ke.items(), 1):
        ty_le = (so_luong / tong_nhan_vien) * 100
        print(f"{i}./Tên đơn vị: {don_vi}.")
        print(f"- Số nhân viên: {so_luong}")
        print(f"- Tỷ lệ: {ty_le:.1f}%.")

thong_ke_nhan_vien()
```
