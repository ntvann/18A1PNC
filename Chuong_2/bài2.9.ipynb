```python
import json

def sorted_dict_to_json():
    python_dict = {
        "zebra": "Dong vat",
        "apple": "Trai cay",
        "book": "Sach",
        "computer": "May tinh"
    }
    
    sorted_json = json.dumps(python_dict, sort_keys=True, indent=4, ensure_ascii=False)
    print("JSON sap xep:")
    print(sorted_json)
    
    return sorted_json

result = sorted_dict_to_json()
```
